```python
df = pd.read_csv("/content/Advanced_pandas_sales_dataset.csv")
print(df.head())
```

```python
print("Display First 10 Rows:", df.head(10))
```

```python
print("Display Dataset Structure:", df.info())
```

```python
print("Find Number of Rows & Columns:", df.shape)
```

```python
print("Display Data Types:", df.dtypes)
```

```python
high_sales = df[df['Sales'] > 3500]

print(high_sales)
```

```python
east_sales = df[
    (df['Region'] == 'East') &
    (df['Sales'] > 2500)
]

print(east_sales)
```

```python
customer_sales = df.groupby('Customer')['Sales'].sum()
customer_sales = customer_sales.sort_values(ascending=False)
print(customer_sales.head(5))
```

```python
region_sales = df.groupby('Region')['Sales'].sum()
print(region_sales)
highest_region = region_sales.idxmax()
print(" ")
print("Highest Revenue Region:", highest_region)
```

```python
product_quantity = df.groupby('Product')['Quantity'].sum()
print(product_quantity)
popular_product = product_quantity.idxmax()
print(" ")
print("Most Popular Product:", popular_product)
```

```python
multi_group = df.groupby(
    ['Region', 'Category']
)['Sales'].sum()
print(multi_group)
```

```python
pivot_table = pd.pivot_table(
    df,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum'
)
print(pivot_table)
print(" ")
print("Region Selling Most Laptops:", pivot_table['Laptop'].idxmax())
```

```python
avg_profit = df.groupby('Category')['Profit'].mean()
print(avg_profit)
most_profitable = avg_profit.idxmax()
print(" ")
print("Most Profitable Category:", most_profitable)
```

```python
discount_profit = df.groupby('Discount')['Profit'].mean()
print(discount_profit)
```

```python
filtered_df = df[df['Sales'] > 3000]
grouped_data = filtered_df.groupby(
    ['Region', 'Product']
)['Sales'].sum()
print(grouped_data)
print(" ")
print("pivot table:")
pivot_result = grouped_data.unstack()
print(pivot_result)
```
