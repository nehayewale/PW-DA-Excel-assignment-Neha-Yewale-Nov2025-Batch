```python
# Load dataset
_df = pd.read_csv('/content/House Price India.csv')

# Display first 5 rows
_df.head()
```

```python
_df.info()
```

```python
_df.describe()
```

```python
# Check missing values
_df.isnull().sum()

# If missing values exist
_df = _df.dropna()
```

```python
average_price = _df['Price'].mean()
median_price = _df['Price'].median()
std_price = _df['Price'].std()

print('Average Price:', average_price)
print('Median Price:', median_price)
print('Standard Deviation:', std_price)
```

```python
most_common_bedrooms = _df['number of bedrooms'].mode()[0]
print(most_common_bedrooms)
```

```python
skewness = _df['Price'].skew()
print(skewness)
```

```python
avg_price_bedrooms = _df.groupby('number of bedrooms')['Price'].mean()
print(avg_price_bedrooms)
```

```python
correlation = _df['Area of the house(excluding basement)'].corr(_df['Price'])
print(correlation)
```

```python
high_price = _df['Price'].quantile(0.95)
low_area = _df['Area of the house(excluding basement)'].quantile(0.25)

anomalies = _df[
    (_df['Price'] > high_price) &
    (_df['Area of the house(excluding basement)'] < low_area)
]

anomalies[['Price', 'Area of the house(excluding basement)', 'number of bedrooms', 'Postal Code']].head()
```

```python
comparison = _df.pivot_table(
    values='Price',
    index='number of floors',
    columns='waterfront present',
    aggfunc='mean'
)

print(comparison)
```

```python
minimum_price = _df['Price'].min()
maximum_price = _df['Price'].max()

print('Minimum Price:', minimum_price)
print('Maximum Price:', maximum_price)
```

```python
highest_location = _df.groupby('Postal Code')['Price'].mean().sort_values(ascending=False)
print(highest_location.head())
```
