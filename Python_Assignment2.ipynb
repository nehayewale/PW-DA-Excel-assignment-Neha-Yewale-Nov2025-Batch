```python
import pandas as pd

df = pd.read_csv("/content/retail_dataset .csv")

# First 5 rows
print(df.head())

# Dataset info
print(df.info())
```

```python
# Missing values per column
print(df.isnull().sum())

# Total missing values
print(df.isnull().sum().sum())
```

```python
# Fill missing values with mean
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())
df['Price'] = df['Price'].fillna(df['Price'].mean())
print(df.isnull().sum())
```

```python
df = df.dropna(subset=['Product Category', 'Region'])
# Display output
print("Dataset after removing missing Product Category and Region:")
print(df)

print("\nShape of dataset (rows, columns):")
print(df.shape)
```

```python
df['Revenue'] = np.array(df['Quantity'] * df['Price'])

# Display updated dataset
print(df.head())
print(df.columns)
```

```python
total_revenue = np.sum(df['Revenue'])
print("Total Revenue:", total_revenue)
```

```python
category_revenue = df.groupby('Product Category')['Revenue'].sum()
print(category_revenue)
```

```python
top3 = category_revenue.sort_values(ascending=False).head(3)

# Bottom 3
bottom3 = category_revenue.sort_values().head(3)

print("Top 3 Categories:\n", top3)
print("\n")
print("Bottom 3 Categories:\n", bottom3)
```

```python
region_revenue = df.groupby('Region')['Revenue'].sum()

# Highest and Lowest
highest_region = region_revenue.idxmax()
lowest_region = region_revenue.idxmin()

print(region_revenue)
print("\n")
print("Highest Revenue Region:", highest_region)
print("\n")
print("Lowest Revenue Region:", lowest_region)
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')
df['Month'] = df['Date'].dt.month
monthly_revenue = df.groupby('Month')['Revenue'].sum()
print(monthly_revenue)
print("\n")
mean_rev = np.mean(df['Revenue'])
median_rev = np.median(df['Revenue'])
std_rev = np.std(df['Revenue'])
print("Mean:", mean_rev)
print("\n")
print("Median:", median_rev)
print("\n")
print("Standard Deviation:", std_rev)
```
